# Credit Risk Analytics & Loan Default Prediction

## Project Overview

This project develops a machine-learning solution for analysing customer credit behaviour and predicting the risk of future default.

The project uses historical customer information including credit limits, repayment status, bill amounts and previous payments to identify patterns associated with credit risk.

## Project Objectives

The main objectives were to:

- explore and understand customer credit data
- identify factors associated with increased default risk
- clean and prepare the dataset for machine learning
- engineer additional credit-risk indicators
- develop predictive classification models
- compare Logistic Regression and Random Forest
- evaluate model performance using multiple metrics
- identify important predictors of customer default

## Machine Learning Models

Two classification models were developed:

1. Logistic Regression
2. Random Forest

The models were evaluated using:

- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC
- Confusion Matrix

## Model Results

| Model | Accuracy | Precision | Recall | F1 Score | ROC-AUC |
|---|---:|---:|---:|---:|---:|
| Logistic Regression | 74.42% | 44.31% | 60.78% | 51.26% | 75.28% |
| Random Forest | **76.72%** | **47.94%** | 60.56% | **53.52%** | **77.57%** |

Random Forest produced the strongest overall performance and was selected as the final model.

## Key Risk Indicators

Random Forest feature importance identified several influential predictors of default, including:

- maximum repayment delay
- recent repayment status
- number of delayed months
- previous repayment behaviour
- credit utilisation
- average payment amount

The results demonstrate that historical repayment behaviour is particularly valuable when assessing customer credit risk.

## Conclusion

The project demonstrates an end-to-end credit-risk analytics workflow covering exploratory data analysis, data preparation, feature engineering, machine learning and model evaluation.

The final Random Forest model achieved approximately **76.7% accuracy** and an **ROC-AUC of 0.78**.

The model should be viewed as a predictive analytical prototype rather than a production lending system. Real-world financial deployment would require further validation, fairness testing, explainability, regulatory review and ongoing model monitoring.
